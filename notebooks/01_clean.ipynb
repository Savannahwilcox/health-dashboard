```python
import pandas as pd 

df = pd.read_csv("../data/places_county_2025.csv")
df.shape
```

```python
cols = [
    "StateAbbr", "StateDesc", "CountyName", "CountyFIPS", "TotalPopulation",
    "DIABETES_CrudePrev", "OBESITY_CrudePrev", "ACCESS2_CrudePrev", "CSMOKING_CrudePrev", 
    "FOODINSECU_CrudePrev", "Geolocation"
]

health = df[cols].copy()
health.head()
```

```python
health.isna().sum()
```

```python
core = ["DIABETES_CrudePrev", "OBESITY_CrudePrev", "ACCESS2_CrudePrev", "CSMOKING_CrudePrev"]
health[core].isna().all(axis=1).sum()
```

```python
health_clean = health.dropna(subset=core).copy()
health_clean.shape
```

```python
health_clean["TotalPopulation"] = health_clean["TotalPopulation"].str.replace(",", "").astype(int)
health_clean.dtypes
```

```python
health_clean.to_csv("../data/places_clean.csv", index=False)
```
